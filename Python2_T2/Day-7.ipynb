# RANDOM FOREST

```python
from sklearn.datasets import load_iris
import pandas as pd
import numpy as np
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split
from sklearn import tree
from sklearn.datasets import load_breast_cancer
from sklearn.metrics import confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
iris=load_iris()
x=iris.data
y=iris.target
# print(x)
# print("--"*50)
# print(y)
```

```python
# test_size=0.2
# random_state=42

x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
RFC=RandomForestClassifier(n_estimators=100,max_depth=4,max_features=3,max_samples=80,criterion='entropy')
```

```python
RFC
```

```python
RFC.fit(x_train,y_train)
```

```python
selected_tree=RFC.estimators_[3]
tree.plot_tree(selected_tree)
```

```python
y_pred=RFC.predict(x_test)
accuracy_score(y_test,y_pred)
```

Random Forest Classifier Implementation

Write Python code to implement a Random Forest Classifier using the following instructions:

Initialize the classifier with:



n_estimators = 100


random_state = 42


Train the model using training data.


Predict class labels for test data.


Evaluate the model by calculating:

```python
data=load_breast_cancer()
x=data.data
y=data.target
```

```python
# test_size=0.2
# random_state=42

x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
RFC=RandomForestClassifier(n_estimators=100,max_depth=4,max_features=3,max_samples=80,criterion='entropy')
```

```python
RFC.fit(x_train,y_train)
```

```python
selected_tree=RFC.estimators_[3]
tree.plot_tree(selected_tree)
```

```python
y_pred=RFC.predict(x_test)
accuracy_score(y_test,y_pred)
```

```python
conf_matrix=confusion_matrix(y_test,y_pred)
conf_matrix
```

```python
sns.heatmap(conf_matrix,annot=True,xticklabels=['A','B'],yticklabels=['A','B'])
plt.xlabel("Predicted")
plt.ylabel("Actual")
```

Diabetes Prediction using Random Forest



You are given a dataset diabetes.csv containing medical records of patients.


The dataset includes predictor variables such as BMI, Age, Insulin, Pregnancies, etc., and a target variable named Outcome.


Your task is to:


Load the dataset using pandas.


Separate independent and dependent variables.


Split dataset into training and testing sets.


Train a Random Forest Classifier.


Predict patient outcomes.


Evaluate performance using:


Confusion Matrix

```python
df=pd.read_csv("Diabetes.csv")
df
```

```python
y=df[['Outcome']]
x=df.drop(columns='Outcome')
```

```python
# test_size=0.2
# random_state=42

x_train,x_test,y_train,y_test=train_test_split(x,y,train_size=0.7,random_state=10)
```

```python
RFC=RandomForestClassifier(n_estimators=500,max_depth=4,max_features=3,max_samples=80,criterion='entropy')
```

```python
RFC.fit(x_train,y_train)
```

```python
selected_tree=RFC.estimators_[3]
tree.plot_tree(selected_tree)
```

```python
x_new=[[8,100,50,29,34,12.5,0.435,25]]
```

```python
y_pred=RFC.predict(x_new)
y_pred
```

```python
if y_pred==0:
    print("No Diabetes")
else:
    print("YES")
```

# SVM ALGORITHM

```python
from sklearn.svm import SVC
```

```python
df=pd.read_csv("Diabetes.csv")
x=df.drop('Outcome',axis=1)
y=df[['Outcome']]
```

```python
# test_size=0.2
# random_state=42

x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
clf=SVC(C=178)
clf.fit(x_train,y_train)
```

```python
y_pred=clf.predict(x_test)
y_pred
```

```python
accuracy_score(y_test,y_pred)
```
