Optimizing Linear Regression Model to Predict Medical Insurance expenses

Objective:

• To build and optimize a Linear Regression model that predicts a
person’s medical insurance expenses based on demographic and health
attributes, ensuring:

• Maximum R2 Score (better prediction accuracy)

• Minimum Mean Squared Error (MSE) (lower prediction error)


Dataset


• File: insurance.csv

Implement

• The goal is to predict medical insurance expenses using a Linear
Regression model based on personal and health attributes like age,
gender, BMI, smoking status, number of children,and region.


• The BMI values are first categorized into four groups: Underweight,
Normal weight, Over-weight, and Obese.


Underweight- > bmi<=18.5


Normal weight -> 18.5 < bmi <= 25


Over-weight -> 26 <= bmi < 30

Obese -> bmi > 30

• test size (0.2) and random states range 0 to 42 are tried to find:

• The model with the highest R2 score (better prediction accuracy)

• The model with the lowest Mean Squared Error (MSE) (lower prediction
error)

• After finding the best models, we predict the insurance expenses for
a new individual with specific given features
(age 34, male, BMI 26, 3 children, smoker, southwest region).

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score,mean_absolute_error
df = pd.read_csv("insurance.csv")
```

```python
print(df)
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
df.duplicated().sum()
```

```python
df["bmi"] = np.where(df.bmi < 18.5, 'Underweight',
np.where(df.bmi < 25, 'Normal weight',
np.where(df.bmi < 30, 'Overweight', 'Obese')))
print(df['bmi'])
```

```python
df = pd.get_dummies(df,drop_first=True)
```

```python
y = df[['charges']]
x = df.drop('charges',axis=1)
```

```python
df
```

```python
states = range(0,43)
mse_value = []
r2_value = []
for i in states:
    x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=i)
    model = LinearRegression()
    model.fit(x_train,y_train)
    y_pred = model.predict(x_test)
    # prediction = pd.DataFrame({
    # 'Actual':y_test.squeeze(),
    #  'Predicted':y_pred.squeeze()
    #  })
    mse = mean_squared_error(y_test,y_pred)
    r2 = r2_score(y_test,y_pred)
    mse_value.append(mse)
    r2_value.append(r2)
print("Mse Values: ")
print(mse_value)
print("R2 Score Values: ")
print(r2_value)
print("Random State : ",mse_value.index(min(mse_value)),"Minimum MSE :",min(mse_value))
print("Random State : ",r2_value.index(max(r2_value)),"Maximum R2 Score : ",max(r2_value))
```

```python
x_test
```

```python
# After finding the best models, we predict the insurance charges for
# a new individual with specific given features (age 34, male, BMI 26, 3 children,
# smoker, southwest region).
x_test = [[34,True,False,True,False,3,True,False,False,True]]
new_y_pred = model.predict(x_test)
print(new_y_pred)
```

#  POLYNOMIAL REGRESSION

```python
from sklearn.preprocessing import PolynomialFeatures
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('Ice_cream selling data.csv')
```

```python
df
```

```python
df.head()
```

```python
x=df[['Temperature (°C)']]
y=df[['Ice Cream Sales (units)']]
```

```python
plt.scatter(x,y)
```

```python
degree=2 #like Power of X no 0f degree=2 the columns will be 3 so no.of degree+1 as constant column is added
poly=PolynomialFeatures(degree=degree)
```

```python
x_poly=poly.fit_transform(x)
x_poly.shape#(49, 3) 
# 3 columns 1.constant 2.x 3.x^2 
```

```python
model=LinearRegression()
model.fit(x_poly,y)
y_pred=model.predict(x_poly)
```

```python
y_pred
```

```python
mean_squared_error(y,y_pred)
```

```python
r2_score(y,y_pred)
```

```python
mean_absolute_error(y,y_pred)
```

```python
plt.scatter(x,y)
plt.plot(x,y_pred,color="red")
```
