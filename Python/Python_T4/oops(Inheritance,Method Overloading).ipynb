# object operator

```python
class Str:
    def __init__(self,mark):
        self.mark=mark
    def __add__(self,other):
        print(self.mark+other.mark)
    def __sub__(self,other):
        print(self.mark-other.mark)
    def __mul__(self,other):
        print(self.mark*other.mark)
    def __eq__(self,other):
        return self.mark==other.mark
    def __le__(self,other):
        return self.mark<=other.mark
    def __gt__(self,other):
        return self.mark>=other.mark
    def __truediv__(self,other):
        return self.mark/other.mark
    def __pow__(self,other):
        return self.mark**other.mark
    def __str__(self):
        return ("Hello")
a=Str(80)
b=Str(70)
print(a+b)
print(a)
print(a*b)
print(a-b)
print(a==b)
print(a<=b)
print(a>b)
a/b
a**b
```

```python
from abc import ABC,abstractmethod
class A:
    @abstractmethod
    def demo(self):
        pass
    def show(self):
        print("num")
class B(A):
    def demo(self):
        print("B")
class C(B):
    def show1(self):
        print("C")
a=B()
a.demo()
a.show()
c=C()
c.show1()

        
    
```

```python
from abc import ABC,abstractmethod
class Employee:
    
    @abstractmethod
    def emp_id(self):
        pass
        
    def emp_det(self):
        self.e_id=int(input("Enter Employee id"))
        self.e_name=input("Enter your name")
        self.e_salary=int(input("Enter Your Salary"))
        
    def print_emp(self):
        print(f"Employee ID : {self.e_id}")
        print(f"Employee Name : {self.e_name}")
        print(f"Employee Basic Salary : {self.e_salary}")
    def ret_sal(self):
        return self.e_salary
class Perks(Employee):
    def calc(self):
        self.DA=0.35*self.e_salary
        self.Hra=0.17*self.e_salary
        self.PF=0.12*self.e_salary
        self.total=self.DA+self.Hra-self.PF
    def print_perks(self):
        print("DA",self.DA)
        print("HRA",self.Hra)
        print("PF",self.PF)
        print("Total Perks",self.total)
class NetSalary(Perks):
    def total_sal(self):
        self.final_sal=self.total+self.e_salary
        
    def print_det(self):
        super().emp_det()
        super().print_emp()
        super().calc()
        super().print_perks()
        self.total_sal()
        print("Salary After Perks",self.final_sal)
a=NetSalary()
a.print_det()


        
    
```
