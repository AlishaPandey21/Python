```python
import numpy as np
scores=[[13,10,9,33],[63,46,90,42],[39,76,13,29],[82,9,29,78],[67,61,59,36]]
x=np.array(scores)
print("Before Adding\n",x)
match_6=[41,87,72,36,92]
y=np.array(match_6)
z=y.reshape(5,1)
x1=np.concatenate((scores,z),axis=1)
print("After Adding 6th match \n",x1)

B_6=[77,83,98,95,89]
B_7=[92,71,52,61,53]
b6=np.array(B_6).reshape(1,5)
b7=np.array(B_7).reshape(1,5)
x5=np.concatenate((x1,b6),axis=0)
x6=np.concatenate((x5,b7),axis=0)
print("After Adding new Batsmen 6 & 7 \n",x5)


sum1=np.sum(x6,axis=1)
print("sum",sum1)

c=sum1.reshape(7,1)
x7=np.concatenate((x6,c),axis=1)
print("Sum Printed After Adding column\n",x7)



max1=np.max(x6[0:,0:-1],axis=1)
print(max1)

plt.plot(max1)
plt.xlabel("Players")
plt.xlabel("Run")
plt.show()

max1=np.max(x6,axis=0)
print(max1)

plt.barh(max1,width=10)
plt.xlabel("Players")
plt.xlabel("Run")
plt.show()


    


```
