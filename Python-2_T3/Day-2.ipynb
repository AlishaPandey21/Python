```python
import requests
from bs4 import BeautifulSoup
```

```python
url="https://www.iplt20.com/auction"
res=requests.get(url)
print(res)
```

```python
soup=BeautifulSoup(res.text,"html.parser")
```

```python
year=soup.find('ul',class_="year-auction nav")
print(year)
```

```python
year.find_all("li")[7].text
```

```python
x=soup.find('table',class_="ih-td-tab w-100 auction-tbl")
rows=x.find_all("tr")
print(rows)
```

```python
columns=[i.text.strip() for i in rows[0].find_all('th')[2:]]
print(columns)
```

```python
data=[]
for i in rows[1:]:
    data.append([j.text.strip() for j in i.find_all('td')[1:]])
print(data)
```

```python
import pandas as pd
df=pd.DataFrame(data=data,columns=columns)
df
```

```python
base_url="https://books.toscrape.com/catalogue/page-{}.html"
titles=[]
prices=[]
ratings=[]
for page in range(1,6):
    url=base_url.format(page)
   # print(url)
    
    res=requests.get(url)
   # print(res)

    soup=BeautifulSoup(res.text,"html.parser")
   # print(soup)
    books=soup.find_all('article',class_="product_pod")
   # print(books)
    for book in books:
        title=book.h3.a['title']
        titles.append(title)

        price=book.find('p',class_="price_color").text
        prices.append(price)
        # print(price)

        rating=book.p['class'][1]
        ratings.append(rating)
        # print(rating)
print(titles)        
```

```python
df=pd.DataFrame({'book_title':titles,
                'book_price':price,
                'book_rating':ratings})
df
```
