```python
import requests
from bs4 import BeautifulSoup
base_url="https://subslikescript.com/movies?page={}"
movie_title=[]
movie_link=[]
for page in range(1,6):
    url=base_url.format(page)
    #print(url)

    res=requests.get(url)
   # print(res)
    soup=BeautifulSoup(res.text,"html.parser")
    #print(soup)

    movie_conatiner=soup.find("ul",class_="scripts-list")
    #print(movie_container)

    movies=movie_conatiner.find_all('a')
   #print(movies)

    for movie in movies:
        title=movie.text.strip()
        movie_title.append(title)


        link="https://subslikescript.com"+movie['href']
        movie_link.append(link)
        

```

```python
website="30 Most Popular Movies Right Now_ What to Watch In Theaters and Streaming __ Rotten Tomatoes – Movie and TV News.html"
file=open(website, encoding="utf-8")
res=file.read()
#print(res)
titles=[]
ratings=[]
soup=BeautifulSoup(res,"html.parser")
#print(soup)

movies=soup.find_all('div',class_="countdown-item")
#print(movies)

for movie in movies[:30]:
    titles.append(movie.h2.a.text)
    ratings.append(movie.find('span',class_="tMeterScore").text)
    
print(len(titles))
```

```python
ratings=soup.find_all('span',class_="tMeterScore")
print(ratings)
print(len(ratings))
```

```python
import pandas as pd
df=pd.DataFrame({
    'titles':titles,
    'ratings':ratings
})
df
```

```python
website="Fact-checks _ PolitiFact.html"
file=open(website,encoding="utf-8")
res=file.read()
soup=BeautifulSoup(res,"html.parser")
statements=[]
sources=[]
dates=[]

articles=soup.find_all("article",class_="m-statement")

for article in articles:
    sources.append(article.find('a',class_="m-statement__name").text.strip())
    statements.append(article.find('div',class_="m-statement__quote").text.strip())
    dates_on=article.find('div',class_="m-statement__desc").text.strip().split('stated on')[1]
    dates.append(dates_on.split('in')[0])
    
```

```python
df=pd.DataFrame({'staements':statements,
                 'sources':sources,
                 "date":dates})
df
```

```python
import numpy as np
website="Certified used Mercedes-Benz for sale.html"
file=open(website,encoding="utf-8")
res=file.read()
soup=BeautifulSoup(res,"html.parser")
name=[]
mileage=[]
price=[]
dealer_name=[]
rating=[]
review_count=[]
# names=soup.find_all('h2',class_="title")
# for name in names:
#     prod_name.append(name.text)
# print(prod_name)
cards=soup.find_all("div",class_="vehicle-card")
for card in cards:
    name.append(card.h2.text)
    mileage.append(card.find('div',class_="mileage").text)
    price.append(card.find('span',class_="primary-price").text)
    try:
        
        rating.append(card.find('span',class_="sds-rating__count").text)
    except:
        rating.append(np.nan)

    dealer_name.append(card.find('div',class_="dealer-name").text.strip())
    review_count.append(card.find('span',class_="sds-rating__link sds-button-link").text)
data=pd.DataFrame({"name":name,
                  "mileage":mileage,
                  "price":price,
                  "rating":rating,
                  "review_count":review_count,
                  "dealer_name":dealer_name,
                  })
data
```
