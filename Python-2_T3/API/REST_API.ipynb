```python
import requests,json
url="http://127.0.0.1:8000/api/players/"
res=requests.get(url,proxies={"http":None ,"https":None})
```

```python
data=res.json()
print(data)
```

```python
json_str=json.dumps(data,indent=5)
print(json_str)
```

```python
data={
    'name':"rohit",
    'runs':1200,
    'innings':230}
res=requests.post(url,json=data,proxies={"http":None ,"https":None})
print(res.status_code)
print(res.json())
```
