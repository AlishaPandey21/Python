```python
url="https://webscraper.io/test-sites/e-commerce/allinone/computers/tablets"
res=requests.get(url)
soup=BeautifulSoup(res.text,"html.parser")
```

```python
name=soup.find_all('a',class_="title")
product_name=[i.text.strip() for i in name]
print(product_name)
```

```python
price=soup.find_all('h4',class_="price float-end card-title pull-right")
prod_price=[i.text.strip() for i in price]
print(prod_price)
```

```python
review=soup.find_all('p',class_="review-count float-end")
prod_review=[i.text.strip() for i in review]
print(prod_review)
```

```python
Desc=soup.find_all('p',class_="description card-text")
prod_Desc=[i.text.strip() for i in Desc]
print(prod_Desc)
```

```python
import pandas as pd
data=pd.DataFrame({'prod_name':product_name,
                  'prod_price':prod_price,
                  'prod_review':prod_review,
                  'prod_dec':prod_Desc})
data
```
